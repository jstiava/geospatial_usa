```python
variable_codes = list(census_codes.keys())
batch_size = 45

batches = [
    variable_codes[i:i + batch_size]
    for i in range(0, len(variable_codes), batch_size)
]
```

```python
BASE_URL = "https://api.census.gov/data/2020/acs/acs5"

dataframes = []

for i, batch in enumerate(batches, start=1):
    print(f"Downloading batch {i} of {len(batches)}...")

    params = {
        "get": ",".join(["NAME", *batch]),
        "for": "place:*",
        "in": "state:17",
        "key": CENSUS_API_KEY
    }


    response = requests.get(
        BASE_URL,
        params=params,
        timeout=120
    )

    print(response.url)
    response.raise_for_status()

    data = response.json()

    df = pd.DataFrame(
        data[1:],
        columns=data[0],
    )

    dataframes.append(df)

df = dataframes[0]

# Merge the remaining batches
for batch_df in dataframes[1:]:

    df = df.merge(
        batch_df,
        on=[
            "state",
            "place",
        ],
        how="left",
        validate="one_to_one",
    )

df.to_csv(
    "IL_2020_ACS5_place_data.csv",
    index=False,
)
```
