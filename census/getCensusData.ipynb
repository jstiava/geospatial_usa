```python
import pandas as pd

url = (
    "https://www2.census.gov/geo/docs/maps-data/data/gazetteer/"
    "2024_Gazetteer/2024_Gaz_counties_national.zip"
)

counties = pd.read_csv(
    url,
    sep="\t",
    dtype={
        "GEOID": str,
        "STATE": str,
        "COUNTY": str,
    },
)

counties["COUNTY_FIPS"] = (
    counties["GEOID"]
    .astype("string")
    .str.zfill(5)
    .str[-3:]
)
```

```python


BASE_URL = "https://api.census.gov/data/2010/acs/acs5"

# Split variables into API-sized batches
variable_codes = list(census_codes.keys())
batch_size = 45

batches = [
    variable_codes[i:i + batch_size]
    for i in range(0, len(variable_codes), batch_size)
]

all_data = []

for _, row in counties[(
    (counties['USPS'] == 'IL')
)].iterrows():
    county = row['COUNTY_FIPS']

    dataframes = []

    for i, batch in enumerate(batches, start=1):
        print(f"Downloading batch {i} of {len(batches)}...")

        params = {
            "get": ",".join(["NAME", *batch]),
            "for": "tract:*",
            "in": f"state:17 county:{county}",
            "key": CENSUS_API_KEY
        }

        response = requests.get(
            BASE_URL,
            params=params,
            timeout=120,
        )
        
        print(response.url)
        response.raise_for_status()

        data = response.json()

        df = pd.DataFrame(
            data[1:],
            columns=data[0],
        )

        dataframes.append(df)


    # Start with the first batch
    df = dataframes[0]

    # Merge the remaining batches
    for batch_df in dataframes[1:]:

        df = df.merge(
            batch_df,
            on=[
                "state",
                "county",
                "tract",
            ],
            how="left",
            validate="one_to_one",
        )

    all_data.append(df)

all_data = pd.concat(
    all_data,
    ignore_index=True,
)

# Create the 11-digit census tract GEOID
all_data["GEOID"] = (
    all_data["state"].str.zfill(2)
    + all_data["county"].str.zfill(3)
    + all_data["tract"].str.zfill(6)
)

all_data.to_csv(
    "IL_2010_ACS5_tract_data.csv",
    index=False,
)

print(f"Saved {len(all_data):,} census tracts")
```
