```python
folder = Path("./census")

IllinoisCensus = {}

for file in folder.iterdir():
    if file.is_file() and file.suffix.lower() == ".csv":
        IllinoisCensus[file.name] = pd.read_csv(file)
```

```python

ILPlaces = gpd.read_file("./maps/raw/IL/tl_2025_17_place/tl_2025_17_place.shp").to_crs(TARGET_CRS)
ILCounties = gpd.read_file("./maps/raw/counties_by_state/IL.geojson").to_crs(TARGET_CRS)
ILPlacesCensusTracts2020  = gpd.read_file('./maps/raw/IL/ILPlacesCensusTracts2020.geojson').to_crs(TARGET_CRS)

ILPlacesCensusTracts2020["GEOID"] = ILPlacesCensusTracts2020["GEOID"].astype(str)
IllinoisCensus['IL_2020_ACS5_tract_data.csv']["GEOID"] = IllinoisCensus['IL_2020_ACS5_tract_data.csv']["GEOID"].astype(str)
```

```python
IllinoisCensus['IL_2020_ACS5_tract_data.csv'] = (
    IllinoisCensus['IL_2020_ACS5_tract_data.csv']
    .drop(
        columns="NAME_y",
        errors="ignore",
    )
    .rename(
        columns={
            "NAME_x": "NAME",
        }
    )
)
```

```python
# Remove all pre-existing census code columns that are na

census_cols_to_remove = [
    code
    for code in census_codes.keys()
    if code != "GEOID"
    and code in ILPlacesCensusTracts2020.columns
]

ILPlacesCensusTracts2020 = (
    ILPlacesCensusTracts2020
    .drop(columns=census_cols_to_remove)
)
```

```python
ILPlacesTracts2020_WithCensusData = ILPlacesCensusTracts2020.merge(
    IllinoisCensus['IL_2020_ACS5_tract_data.csv'],
    how="left",
    on='GEOID'
)
```

```python
tracts_with_no_place = ILPlacesTracts2020_WithCensusData[(
    (ILPlacesTracts2020_WithCensusData['PLACE_GEOID'].isna())
    | (ILPlacesTracts2020_WithCensusData['PLACE_NAME'].isna())
)]

tracts_with_a_place= ILPlacesTracts2020_WithCensusData[(
    (~ILPlacesTracts2020_WithCensusData['PLACE_GEOID'].isna())
    & (~ILPlacesTracts2020_WithCensusData['PLACE_NAME'].isna())
)]
```

```python
tracts_with_no_place
# tracts_with_no_place.to_file("WithNoPlace_ILPlacesTracts2020_WithCensusData.geojson")
```

```python
tracts_with_a_place = tracts_with_a_place[(
    tracts_with_a_place['PLACE_NAME'] != 'Chicago'
)]
```

```python
tracts_with_a_place = (
    tracts_with_a_place
    .drop(
        columns="NAME_x",
        errors="ignore",
    )
    .rename(
        columns={
            "NAME_y": "NAME",
        }
    )
)
```

```python
# Census codes that are medians, not counts
median_codes = {
    "B19013_001E",  # Median Household Income
    "B25064_001E",  # Median Gross Rent
}

# GEO_ID is an identifier, not something to aggregate
excluded_codes = {
    "GEOID",
}

# Sum all count variables that actually exist in the GeoDataFrame
sum_cols = [
    code
    for code in census_codes.keys()
    if code in tracts_with_a_place.columns
    and code not in median_codes
    and code not in excluded_codes
]

# Keep only median codes that exist
median_cols = [
    code
    for code in median_codes
    if code in tracts_with_a_place.columns
]

# Create the aggregation dictionary
aggfunc = {
    **{col: "sum" for col in sum_cols},
    **{col: "median" for col in median_cols},
}

ILPlacesRebuiltWithTractOverlap_Census2020 = tracts_with_a_place.dissolve(
    by=["PLACE_GEOID", "PLACE_NAME"],
    aggfunc=aggfunc,
    as_index=False,
    dropna=False,
)
```

```python
def ZoomToTheExtentOf(ax, geoDataFrame):
    minx, miny, maxx, maxy = geoDataFrame.total_bounds

    ax.set_xlim(minx, maxx)
    ax.set_ylim(miny, maxy)

    return ax
```

```python
fig, ax = plt.subplots(figsize=(14, 14))


ILPlacesRebuiltWithTractOverlap_Census2020.plot(
    ax=ax,
    column="PLACE_GEOID",
    categorical=True,
    legend=False,
    edgecolor="#00000050",
    linewidth=1,
)



ILPlaces.plot(
    ax=ax,
    linewidth=1,
    color="#ffffff00",
    edgecolor="#000"
)

ZoomToTheExtentOf(ax, ILCounties[(
    (ILCounties['NAME'].isin(['COOK', 'KANE', 'LAKE', 'WILL']))
)])

ax.set_axis_off()
ax.set_title("Illinois Places — Total Population")
plt.show()
```
