```python
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import pygris 

TARGET_CRS = "EPSG:5070"
```

```python
# Get the cartographic boundary file with cb = True
cartographic = pygris.counties(year=1990, cb = True, cache = True)

# Plot the two side-by-side to compare them
fig, ax = plt.subplots(ncols = 2)
cartographic.plot(ax = ax[0])
cartographic.plot(ax = ax[1])

ax[0].set_title("TIGER/Line")
ax[1].set_title("Cartographic")
```

```python
cartographic.to_file("USCounties_1990.geojson")
```
